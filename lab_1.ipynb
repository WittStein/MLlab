## Статистический разбор данных

### **Анализ:**

1. hour_of_day - Отношений (количественная шкала)
часы дня (1-24/0-23) имеют равные промежутки, абсолютный 0 - начало дня.
2. cash_type - Номинальная (качественная шкала) это просто метки, нельзя сказать что кто то больше или меньше.
3. money - Отношений (количественная шкала) есть истинный 0, где это полное отсутствие денег
4. coffee_name	- Номинальная (качественная шкала) это просто метки, нельзя сказать что кто то больше или меньше.
5. Time_of_Day	- Порядковая (качественная шкала) можно сказать что какое то значение больше(позже, раньше), измерить нельзя
6. Weekday	- Порядковая (качественная шкала) можно сказать что какое то значение больше(позже, раньше), измерить нельзя
7. Month_name	- Порядковая (качественная шкала) можно сказать что какое то значение больше(позже, раньше), измерить нельзя
8. Weekdaysort	- Порядковая (количественная шкала) числовое представление дней, числа здесь используются для обозначения порядка, а не для арифметических операций (Вт - Пн ≠ 1 дню в абсолютном смысле для всех данных, нет уточнения в какой час дня).
9. Monthsort	- Порядковая (количественная шкала) числовое представление месяцев, числа здесь используются для обозначения порядка, а не для арифметических операций
10. Date	- Интервальная (количественная шкала) интервалы равны между днями, нет абсолютного 0 (например 31.12.2019-01.01.2020 разница 1 день, нет абсолютного 0)
11. Time	- Интервальная (количественная шкала) интервалы равны между секундами, нет абсолютного 0 (например 23:59:59-00:00:00 разница 1 секунда, нет абсолютного 0)

абсолютный 0 во времени можно наблюдать у таймера, когда у него стоит отметка 0 минут 0 секунд, что означает полное отсутсвие времени. Абсолютный 0 у даты: таковым можно счить точку отсчета начала вселенной

## **4. Визуализация распределения переменных**


1. **Гистограмма**

```python
sns.histplot(df['money'], kde=True)
plt.title('Распределение money')
plt.xlabel('money')
plt.ylabel('Частота')
plt.show()
```

1. Данное распределение не является нормальным. Данные не распределены равномерно по обе стороны от центрального пика. Гистограмма и кривая плотности показывают неровный, волнистый вид, не похожий на ровный колокол.
2. Данное распределение имеет один ярко выраженный пик. Можно заметить что кофе чаще всего покупали от 30-40 денежных единиц. Данные не являются однородными, существует две основные группы: одна с меньшей суммой денег (до 25), которая является немногочисленной, и другая (от 25 до 40), которая является многочисленной. Суммы около 20 единиц встречаются крайне редко.

2. **Остальные гистограммы**

```python
num_cols = df.columns.intersection(['hour_of_day', 'Date', 'Time'])

fig, axes = plt.subplots(1, len(num_cols), figsize=(6 * len(num_cols), 4))

for ax, col in zip(axes, num_cols):
    sns.histplot(df[col], kde=True, ax=ax)
    ax.set_title(f'Распределение {col}')
    ax.set_xlabel(col)
    ax.set_ylabel('Частота')

plt.tight_layout()
plt.show()
```

Первые два графика отличны от нормы(распределения не симметричны).
1. Первый график: Наибольшая частота наблюдается в период с 10 до 11 часов, где столбец достигает максимальной высоты (около 325).
Второй крупный пик приходится на период с 16 до 17 часов.
Третий пик около 22 часов.
2. Второй график также отличен от нормы, но обычно к графикам даты не применяется анализ "нормальности", Здесь анализируется тренды, сезонность.
Есть много острых пиков, где частота достигает 25-27. Это указывает на конкретные даты с чрезвычайно высокой активностью.
3. Третий график: Распределение отлично от нормального. Оно очень близко к равномерному распределению, ограниченному некоторыми пределами. у Time слишком много уникальных значений, повторов нет скорее всего

3. **Диаграмма по категориальным переменным:**

```python
plt.figure(figsize=(15, 6))
sns.countplot(x='coffee_name', data=df)
plt.title('Распределение coffee_name')
plt.xlabel('coffe_name')
plt.ylabel('Количество')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

Данный график показывает частоту (Количество) категории (coffee_name). Столбец отражает, сколько раз был выбран каждый вид кофе.

Различия в высоте столбцов отражают лишь популярность или спрос на определенный вид кофе, а не аномалию в числовом измерении.

```python
num_cols = df.columns.intersection(["Time_of_Day", "Month_name", "Weekdaysort", "Monthsort"])

fig, axes = plt.subplots(1, len(num_cols), figsize=(6 * len(num_cols), 4))

for ax, col in zip(axes, num_cols):
    sns.countplot(x=col, data=df, ax=ax)
    ax.set_title(f'Распределение {col}')
    ax.set_xlabel(col)
    ax.set_ylabel('Количество')

plt.tight_layout()
plt.show()
```

1. Первый график: тип распределения почти равномерный. Выбросов нет. Это распределение частот по категориям.
2. Второй график:  Здесь мы видим частоту (количество) для каждой категории (месяца). Низкая частота в январе или апреле не является выбросом, она указывает на сезонный спад активности.
3. Третий график: тип распределения почти равномерный. Выбросов нет. Это распределение частот по категориям.
4. Четвертый график: Это график частот категорий. Различия в высоте столбцов отражают сезонность, а не аномальные числовые значения.

**Корреляционная матрица:**

```python
numeric_df = df.select_dtypes(include=['number'])

correlation_matrix = numeric_df.corr()

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Корреляционная матрица')
plt.show()
```

1. На основе этого набора числовых переменных, самая сильная связь  наблюдается между часом покупки и суммой покупки (0.2).
2. Практически нулевая корреляция. То, в какой день недели совершается покупка, не связано с часом покупки.
3. Очень слабая положительная связь. Месяц года почти не влияет на час покупки.
4. Очень слабая отрицательная связь. День недели почти не влияет на сумму покупки.
5. Очень слабая положительная связь. Месяц сортировки не влияет на день сортировки.
